# Drills — the whiteboard problems, on random numbers

The other notebooks build understanding. This one builds **fluency**, which is a
different thing and the one that fails under pressure.

Every drill generates fresh numbers each run, so what you rehearse is the method
rather than one memorised example. Each has a target time. The target is not
speed for its own sake — it is being able to **narrate while computing**, because
an interviewer is listening to your reasoning and silence reads as uncertainty.

### How to use it

1. Run a drill cell. Read the question.
2. Work it **on paper, out loud**, with a timer. No autocomplete, no calculator
   for the rounding.
3. Call `.reveal()` and compare — not just the number, the *steps*.
4. Rerun the cell for a fresh instance. Repeat until the shape is automatic.

The six drills, and what each one is really testing:

| Drill | Tests |
|---|---|
| 1 KV cache sizing | Do you use KV heads or attention heads? (the GQA discount) |
| 2 Decode ceiling | Do you know decode is bandwidth bound, and can you show it? |
| 3 TTFT | Can you switch regimes — prefill is the one place FLOPs matter |
| 4 Concurrency | Can you turn memory into a capacity number in one pass? |
| 5 Serving cost | Can you price a workload live, including the routing option? |
| 6 QLoRA OOM | Can you budget training memory and name the term that kills it? |
